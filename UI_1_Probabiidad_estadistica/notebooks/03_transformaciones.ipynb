```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head(10)
```

```python
df.rename(columns={
    "entity": "Pais",
    "code": "codigo",
    "year": "año",
    "life_expectancy_0": "esperanzaVida"
    }, inplace=True)
```

```python
df.head(10)
```

```python
df.set_index("Pais", inplace=True)
```

```python
df.set_index(["entity", "year"], inplace=True)
```

```python
df.loc[df["entity"].str.contains("UNI", case=False, regex=True)]
```

```python
df.dropna(subset=["code"], inplace=True)
df.isnull().sum()
```

```python
df.head()
```

```python
resultado = (
    df.groupby("entity")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=True)
)
```

```python
resultado.head(5)
```

```python
df_2020 = df[df["year"] == 2020]
```

```python
df_2020.head(5)
```

```python
resultado_2020 = (
    df_2020.groupby("entity")["life_expectancy_0"]
    .min()
    .sort_values(ascending=True)
)
```
